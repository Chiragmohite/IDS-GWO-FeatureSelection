# Model Evaluation
Compares the baseline (all-features) model against the GWO-selected-feature model. Run `python main.py` first to generate `outputs/reports/comparison_table.csv`.

```
import sys, os
sys.path.append(os.path.abspath('..'))
import pandas as pd
import config

comparison = pd.read_csv(os.path.join(config.REPORTS_DIR, 'comparison_table.csv'))
comparison
```

## Visual comparison

```
import matplotlib.pyplot as plt

metrics = ['accuracy', 'precision', 'recall', 'f1_score']
comparison.set_index('model')[metrics].plot(kind='bar', figsize=(9,5))
plt.ylabel('Score (%)')
plt.title('Baseline vs GWO-Selected Feature Model')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
